```python
import numpy as np
import scipy.stats as stats

data = [1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85,
        1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29]

n = len(data)
mean = np.mean(data)
std = np.std(data, ddof=1)

# (a) t-distribution
t_value = stats.t.ppf(0.995, df=n-1)
margin_t = t_value * (std / np.sqrt(n))
ci_t = (mean - margin_t, mean + margin_t)

# (b) z-distribution
z_value = 2.576
sigma = 0.2
margin_z = z_value * (sigma / np.sqrt(n))
ci_z = (mean - margin_z, mean + margin_z)

print("Mean:", mean)
print("99% CI (t):", ci_t)
print("99% CI (z):", ci_z)
```
