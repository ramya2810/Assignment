```python
import math

# Given values
x_bar = 3050
mu = 1000 + 5 * 600
sigma = 5 * 25
n = 25

# Z calculation
z = (x_bar - mu) / (sigma / math.sqrt(n))

print("Z value:", z)

# Critical value
z_critical = 1.645

if z > z_critical:
    print("Reject Null Hypothesis")
else:
    print("Fail to Reject Null Hypothesis")
```
