```python
# Create list
lst = [1, 2, 3, 4, 5]

# append
lst.append(6)
print("After append:", lst)

# extend
lst.extend([7, 8])
print("After extend:", lst)

# insert
lst.insert(2, 99)
print("After insert:", lst)

# remove
lst.remove(99)
print("After remove:", lst)

# pop
lst.pop()
print("After pop:", lst)

# index
print("Index of 3:", lst.index(3))

# count
print("Count of 2:", lst.count(2))

# sort
lst.sort()
print("After sort:", lst)

# reverse
lst.reverse()
print("After reverse:", lst)

# clear
lst.clear()
print("After clear:", lst)
```

```python
#Dictionary
person = {
    "name": "John",
    "age": 21,
    "city": "Chennai"
}

# Update age
person["age"] = 22

# Add email
person["email"] = "john@email.com"

print(person)
```

```python
#Eligibility to vote
age = int(input())

if age >= 18:
    print("Eligible to vote")
else:
    print("Not eligible to vote")
```

```python
#Grading System
marks = int(input())
if marks >= 90:
    print("A")
elif marks >= 75:
    print("B")
elif marks >= 50:
    print("C")
else:
    print("Fail")
```

```python
#Checking Number
num = int(input())
if num > 0:
    if num % 2 == 0:
        print("Positive and Even")
    else:
        print("Positive but Odd")
elif num == 0:
    print("Zero")
else:
    print("Negative")
```
