```python
import pandas as pd
import numpy as np

# Create DataFrame (10 rows)
data = {
    "Name": ["Charlie", "Andres", "John", "Robert", "Tom", "Aries", "Ash", "Felix", "Jasper", "Keifer"],
    "Age": [23, 25, 22, 24, 23, 26, 25, 24, 22, 23],
    "City": ["Chennai", "Delhi", "Mumbai", "Chennai", "Bangalore", "Delhi", "Mumbai", "Chennai", "Bangalore", "Delhi"],
    "Salary": [30000, 40000, np.nan, 35000, 30000, np.nan, 45000, 30000, 40000, 35000]
}

df = pd.DataFrame(data)

print("Original DataFrame:\n", df)

# 1. info() and describe()
print("\n--- INFO ---")
df.info()

print("\n--- DESCRIBE ---")
print(df.describe())

# 2. Select 'Name' and 'City'
print("\n--- Selected Columns (Name & City) ---")
print(df[["Name", "City"]])

# 3. Drop 'City' column
df = df.drop(columns=["City"])
print("\n--- After Dropping City ---")
print(df)

# 4. Fill missing Salary with mean (BEST PRACTICE)
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())
print("\n--- After Filling Missing Salary ---")
print(df)

# 5. Remove duplicate rows
df = df.drop_duplicates()
print("\n--- After Removing Duplicates ---")
print(df)
```
